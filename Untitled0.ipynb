```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.preprocessing import StandardScaler

# 1. Reload raw data fresh to avoid KeyError from re-running
df = pd.read_csv("/Titanic-Dataset.csv")

# 2. Handle missing values
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])

# Drop non-predictive columns
cols_to_drop = [c for c in ["Cabin", "PassengerId", "Name", "Ticket"] if c in df.columns]
df.drop(columns=cols_to_drop, inplace=True)

# 3. Encode categorical variables
df["Sex"] = df["Sex"].map({"male": 0, "female": 1})
df = pd.get_dummies(df, columns=["Embarked"], drop_first=True)

# 4. Outlier detection & visualization (Boxplots)
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.boxplot(ax=axes[0], y=df["Fare"], color="skyblue")
axes[0].set_title("Fare Outliers (Before Removal)")

# Outlier Removal using IQR method
Q1 = df["Fare"].quantile(0.25)
Q3 = df["Fare"].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

df_clean = df[(df["Fare"] >= lower_bound) & (df["Fare"] <= upper_bound)].copy()

sns.boxplot(ax=axes[1], y=df_clean["Fare"], color="lightgreen")
axes[1].set_title("Fare Outliers (After Removal)")
plt.tight_layout()
plt.show()

# 5. Feature scaling (Standardization)
scaler = StandardScaler()
df_clean[["Age", "Fare"]] = scaler.fit_transform(df_clean[["Age", "Fare"]])

# 6. Display clean dataset
print("\n--- CLEANED & SCALED DATASET HEAD ---")
print(df_clean.head())
```
